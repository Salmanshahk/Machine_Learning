```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Actual Values
y_true = [2, 4, 6, 8]

# Predicted Values
y_pred = [3, 5, 7, 9]

mse = mean_squared_error(y_true, y_pred)
mae = mean_absolute_error(y_true, y_pred)
```

```python
# Actual Values
y_true = [2, 4, 6, 8]

# Predicted Values
y_pred = [3, 5, 7, 9]

# MSE
mse = mean_squared_error(y_true, y_pred)
print("MSE:", mse)

# MAE
mae = mean_absolute_error(y_true, y_pred)
print("MAE:", mae)

# RMSE
rmse = np.sqrt(mse)
print("RMSE:", rmse)
```
